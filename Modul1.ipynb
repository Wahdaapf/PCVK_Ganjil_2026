```python
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow # for image display
from skimage import io
from skimage import transform
from PIL import Image
import matplotlib.pylab as plt
```

```python
urls = ["https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg",
        "https://iiif.lib.ncsu.edu/iiif/0016007/full/800,/0/default.jpg",
        "https://placekitten.com/800/571"]

for url in urls:
    image = io.imread(url)
    image = cv.resize(image, (0,0), fx=0.5, fy=0.5)
    image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
    final_frame = cv.hconcat((image, image_2))
    cv2_imshow(final_frame)
    print('\n')
```

```python
tinggi = image_2.shape[0]
lebar = image_2.shape[1]
print("resolusi image: tinggi x lebar = ", tinggi, " x ", lebar)
cv2_imshow(image_2)
```

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

#membuat garis horizontal ditengah image
for y in range (lebar):
    image_3[int((tinggi)/2),y] = [255,255,255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# membuat garis vertikal di tengah image
for y in range(tinggi):
    image_3[y, int(lebar/2)] = [255,255,255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# membuat garis diagonal dari kiri atas ke kanan bawah
for y in range(tinggi):
    x = int((lebar / tinggi) * y)
    if x < lebar:
        image_3[y, x] = [255,255,255]

# membuat garis diagonal dari kanan atas ke kiri bawah
for y in range(tinggi):
    x = int(lebar - (lebar / tinggi) * y) - 1
    if 0 <= x < lebar:
        image_3[y, x] = [255,255,255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# membuat garis horizontal putih di tengah image
y = int(tinggi / 2)

panjang = 100
x_awal = int((lebar - panjang) / 2)
x_akhir = int((lebar + panjang) / 2)

for x in range(x_awal, x_akhir):
    image_3[y, x] = [255,255,255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# membuat kotak berwarna putih
x_awal = 100
x_akhir = 200
y_awal = 100
y_akhir = 200

for y in range(y_awal, y_akhir):
    for x in range(x_awal, x_akhir):
        image_3[y, x] = [255,255,255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

```python
import cv2 as cv
import matplotlib.pyplot as plt

img = cv.imread('/content/drive/MyDrive/Google Collab/Citra/Modul2/kucing.webp')

tinggi, lebar = img.shape[:2]

# Rectangle di sekitar wajah
pt1 = (int(lebar * 0.30), int(tinggi * 0.20))
pt2 = (int(lebar * 0.70), int(tinggi * 0.65))

cv.rectangle(
    img,
    pt1,
    pt2,
    color=(0, 255, 0),
    thickness=max(2, int(min(lebar, tinggi) * 0.01))
)

# Circle di sekitar wajah
center = (
    int(lebar * 0.50),
    int(tinggi * 0.40)
)

radius = int(min(lebar, tinggi) * 0.18)

cv.circle(
    img,
    center=center,
    radius=radius,
    color=(255, 0, 0),
    thickness=max(2, int(min(lebar, tinggi) * 0.01))
)

img_rgb = cv.cvtColor(img, cv.COLOR_BGR2RGB)

plt.figure(figsize=(8, 8))
plt.imshow(img_rgb)
plt.title("Rectangle dan Circle pada Wajah")
plt.axis("off")
plt.show()
```

```python
import cv2 as cv
import matplotlib.pyplot as plt

img = cv.imread('/content/drive/MyDrive/Google Collab/Citra/Modul2/kucing.webp')
img_rgb = cv.cvtColor(img, cv.COLOR_BGR2RGB)

tinggi, lebar = img_rgb.shape[:2]

# Rectangle di sudut bawah kiri
pt1 = (int(lebar * 0.05), int(tinggi * 0.70))
pt2 = (int(lebar * 0.35), int(tinggi * 0.95))

cv.rectangle(
    img_rgb,
    pt1,
    pt2,
    color=(255, 0, 0),
    thickness=max(2, int(min(lebar, tinggi) * 0.01))
)

# Menambahkan nama file
font = cv.FONT_HERSHEY_SIMPLEX

cv.putText(
    img_rgb,
    text='kucing.webp',
    org=(int(lebar * 0.05), int(tinggi * 0.97)),
    fontFace=font,
    fontScale=max(0.5, min(lebar, tinggi) / 500),
    color=(255, 255, 0),
    thickness=max(1, int(min(lebar, tinggi) / 300)),
    lineType=cv.LINE_AA
)

plt.figure(figsize=(8, 6))
plt.imshow(img_rgb)
plt.title("Citra dengan Nama File")
plt.axis("off")
plt.show()
```

```python
import cv2 as cv
import matplotlib.pyplot as plt

# Membaca gambar KTM
img = cv.imread('/content/drive/MyDrive/Google Collab/Citra/Modul1/kartu-pelajar.jpeg')

# Mengetahui ukuran gambar
tinggi, lebar = img.shape[:2]

print("Lebar  :", lebar)
print("Tinggi :", tinggi)

# =========================================================
# 1. MENUTUP NAMA
# =========================================================

cv.rectangle(
    img,
    (int(lebar * 0.21), int(tinggi * 0.70)),
    (int(lebar * 0.60), int(tinggi * 0.80)),
    (0, 0, 0),
    -1
)

# =========================================================
# 2. MENUTUP NIM
# =========================================================

cv.rectangle(
    img,
    (int(lebar * 0.21), int(tinggi * 0.81)),
    (int(lebar * 0.45), int(tinggi * 0.87)),
    (255, 255, 255),
    -1
)

# =========================================================
# 3. MENUTUP PRODI
# =========================================================

cv.rectangle(
    img,
    (int(lebar * 0.21), int(tinggi * 0.87)),
    (int(lebar * 0.55), int(tinggi * 0.94)),
    (0, 255, 255),
    -1
)

# Konversi BGR ke RGB untuk Matplotlib
img_rgb = cv.cvtColor(img, cv.COLOR_BGR2RGB)

# Menampilkan hasil
plt.figure(figsize=(12, 8))
plt.imshow(img_rgb)
plt.title("KTM - Kelompok 1 (Nama, NIM, dan Prodi Ditutup)")
plt.axis("off")
plt.show()
```
