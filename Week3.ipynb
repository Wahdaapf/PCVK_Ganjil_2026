```python
import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow

print('Mengubah tingkat kecerahan citra dengan Transformasi Log')
print('-------------------------------------------------------')

try:
    brightness = int(input('Masukkan nilai kecerahan: '))
except ValueError:
    print('Error, input harus berupa angka')
    brightness = 50

# Membaca gambar
original = cv.imread('/content/drive/MyDrive/Colab Notebooks/rumah.png')

# Transformasi logaritmik
log_image = np.zeros(original.shape, dtype=np.float32)

for y in range(original.shape[0]):
    for x in range(original.shape[1]):
        for c in range(original.shape[2]):
            r = original[y, x, c]

            # Rumus transformasi logaritmik
            s = brightness * np.log(1 + r)

            # Membatasi nilai piksel ke 0-255
            log_image[y, x, c] = np.clip(s, 0, 255)

# Ubah ke uint8 agar bisa ditampilkan
log_image = log_image.astype(np.uint8)

# Menampilkan gambar asli dan hasil transformasi
final_frame = cv.hconcat((original, log_image))
cv2_imshow(final_frame)
```

```python
import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow

# Membaca gambar
original = cv.imread('/content/drive/MyDrive/Colab Notebooks/rumah.png')

if original is None:
    print("Gambar tidak ditemukan. Periksa kembali path gambar.")
else:

    height, width = original.shape[:2]

    # =========================================
    # A. AVERAGING
    # Rumus: (R + G + B) / 3
    # =========================================
    averaging = np.zeros((height, width), dtype=np.uint8)

    for y in range(height):
        for x in range(width):

            B = int(original[y, x, 0])
            G = int(original[y, x, 1])
            R = int(original[y, x, 2])

            averaging[y, x] = (R + G + B) / 3

    # Ubah grayscale menjadi 3 channel
    averaging_3ch = cv.cvtColor(averaging, cv.COLOR_GRAY2BGR)


    # =========================================
    # B. LIGHTNESS
    # Rumus: (max(R,G,B) + min(R,G,B)) / 2
    # =========================================
    lightness = np.zeros((height, width), dtype=np.uint8)

    for y in range(height):
        for x in range(width):

            B = int(original[y, x, 0])
            G = int(original[y, x, 1])
            R = int(original[y, x, 2])

            maksimum = max(R, G, B)
            minimum = min(R, G, B)

            lightness[y, x] = (maksimum + minimum) / 2

    # Ubah grayscale menjadi 3 channel
    lightness_3ch = cv.cvtColor(lightness, cv.COLOR_GRAY2BGR)


    # =========================================
    # C. LUMINANCE
    # Rumus: 0.299R + 0.587G + 0.114B
    # =========================================
    luminance = np.zeros((height, width), dtype=np.uint8)

    for y in range(height):
        for x in range(width):

            B = float(original[y, x, 0])
            G = float(original[y, x, 1])
            R = float(original[y, x, 2])

            luminance[y, x] = (
                0.299 * R +
                0.587 * G +
                0.114 * B
            )

    # Ubah grayscale menjadi 3 channel
    luminance_3ch = cv.cvtColor(luminance, cv.COLOR_GRAY2BGR)


    # =========================================
    # MENAMPILKAN HASIL
    # =========================================

    print("a. Averaging")
    frame_averaging = cv.hconcat((original, averaging_3ch))
    cv2_imshow(frame_averaging)

    print("b. Lightness")
    frame_lightness = cv.hconcat((original, lightness_3ch))
    cv2_imshow(frame_lightness)

    print("c. Luminance")
    frame_luminance = cv.hconcat((original, luminance_3ch))
    cv2_imshow(frame_luminance)
```

```python
import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow

# Membaca gambar
original = cv.imread('/content/drive/MyDrive/Colab Notebooks/rumah.png')

if original is None:
    print("Gambar tidak ditemukan. Periksa path gambar.")
else:

    # Membuat salinan gambar
    hasil = original.copy()

    # Memproses setiap pixel
    for y in range(original.shape[0]):
        for x in range(original.shape[1]):

            # OpenCV: BGR
            B = int(original[y, x, 0])
            G = int(original[y, x, 1])
            R = int(original[y, x, 2])

            # Mendeteksi warna merah
            if R > 100 and R > G * 1.2 and R > B * 1.2:
                # Tetap pertahankan warna merah
                hasil[y, x] = original[y, x]

            else:
                # Warna lain diubah menjadi grayscale
                gray = int(0.299 * R + 0.587 * G + 0.114 * B)

                hasil[y, x] = [gray, gray, gray]

    # Menggabungkan gambar asli dan hasil
    final_frame = cv.hconcat((original, hasil))

    # Menampilkan
    print("Gambar asli                         Hasil")
    cv2_imshow(final_frame)
```

```python
import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow

print('Gamma Correction pada citra')
print('--------------------------------')

try:
    gamma = float(input('Masukkan nilai Gamma: '))
except ValueError:
    print('Error, not a number')
    gamma = 1.0

# Membaca gambar
original = cv.imread('/content/drive/MyDrive/Colab Notebooks/rumah.png')

if original is None:
    print('Gambar tidak ditemukan. Periksa kembali path gambar.')
else:

    # Membuat citra hasil
    gamma_image = np.zeros_like(original, dtype=np.uint8)

    # Proses Gamma Correction
    for y in range(original.shape[0]):
        for x in range(original.shape[1]):
            for c in range(original.shape[2]):

                r = original[y, x, c]

                # Normalisasi nilai piksel 0-255 menjadi 0-1
                normalized = r / 255.0

                # Rumus Gamma Correction
                corrected = 255 * (normalized ** gamma)

                # Simpan hasil
                gamma_image[y, x, c] = np.clip(corrected, 0, 255)

    # Menggabungkan gambar asli dan hasil
    final_frame = cv.hconcat((original, gamma_image))

    print()
    print('Gambar asli                     Gamma Correction')
    cv2_imshow(final_frame)
```

```python
import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow

# Membaca citra grayscale
original = cv.imread(
    '/content/drive/MyDrive/Colab Notebooks/rumah.png',
    cv.IMREAD_GRAYSCALE
)

# Meminta bit depth dari user
try:
    bit_depth = int(input('Masukkan bit depth tujuan (1-8): '))

    if bit_depth < 1 or bit_depth > 8:
        print('Bit depth harus antara 1 sampai 8.')
    else:

        # Bit depth awal
        original_bit_depth = 8

        # Jumlah level warna
        jumlah_level = pow(2, bit_depth)

        # Jarak antar level grayscale
        level = 255 / (jumlah_level - 1)

        # Membuat citra hasil
        depth_image = np.zeros(original.shape, original.dtype)

        # Proses kuantisasi
        for y in range(original.shape[0]):
            for x in range(original.shape[1]):

                # Nilai pixel asli
                pixel = original[y, x]

                # Menentukan level terdekat
                index = round(pixel / level)

                # Mengubah ke nilai grayscale baru
                new_pixel = index * level

                # Menyimpan hasil
                depth_image[y, x] = np.clip(new_pixel, 0, 255)

        # Informasi
        print()
        print(f'Bit depth awal   : {original_bit_depth} bit')
        print(f'Bit depth tujuan : {bit_depth} bit')
        print(f'Jumlah level     : {jumlah_level}')
        print(f'Nilai level      : {level:.2f}')

        # Menampilkan gambar
        final_frame = cv.hconcat((original, depth_image))

        print()
        print('Grayscale 8-bit              Grayscale hasil')
        cv2_imshow(final_frame)

except ValueError:
    print('Error, input harus berupa angka.')
```

```python
import cv2 as cv
import numpy as np
import os

# ==========================================
# Membaca gambar asli
# ==========================================

original = cv.imread(
    '/content/drive/MyDrive/Colab Notebooks/rumah.png'
)

if original is None:
    print('Gambar galaxy.jpg tidak ditemukan!')
else:
    print('Gambar asli berhasil dibaca.')

# ==========================================
# Membuat folder noises
# ==========================================

noise_folder = '/content/drive/MyDrive/Polinema/Kuliah/PCVK/Images/noises'

os.makedirs(noise_folder, exist_ok=True)

# ==========================================
# Membuat 100 gambar Gaussian Noise
# ==========================================

np.random.seed(42)

for i in range(100):

    # Membuat Gaussian Noise
    noise = np.random.normal(
        loc=0,
        scale=25,
        size=original.shape
    )

    # Menambahkan noise ke gambar asli
    noisy_image = original.astype(np.float32) + noise

    # Batasi nilai pixel 0-255
    noisy_image = np.clip(
        noisy_image,
        0,
        255
    ).astype(np.uint8)

    # Simpan sebagai JPG
    filename = os.path.join(
        noise_folder,
        f'noise_{i+1:03d}.jpg'
    )

    cv.imwrite(filename, noisy_image)

print('100 gambar Gaussian Noise berhasil dibuat.')
```

```python
import cv2 as cv
import numpy as np
import glob
from google.colab.patches import cv2_imshow

# Membaca gambar asli
original = cv.imread(
    '/content/drive/MyDrive/Colab Notebooks/rumah.png'
)

# Membaca semua gambar noise
cv_img = []

for img in sorted(glob.glob(
    '/content/drive/MyDrive/Polinema/Kuliah/PCVK/Images/noises/*.jpg'
)):
    n = cv.imread(img)

    if n is not None:
        cv_img.append(n)

print('Jumlah citra noise:', len(cv_img))

# Jumlah gambar yang akan dirata-rata
jumlah_citra = [10, 20, 40, 80, 100]

hasil = []

for jumlah in jumlah_citra:

    # Menjumlahkan semua citra
    average_image = np.zeros(
        original.shape,
        dtype=np.float32
    )

    for i in range(jumlah):
        average_image += cv_img[i].astype(np.float32)

    # Rata-rata
    average_image /= jumlah

    # Ubah ke uint8
    average_image = np.clip(
        average_image,
        0,
        255
    ).astype(np.uint8)

    # PSNR
    psnr = cv.PSNR(
        original,
        average_image
    )

    hasil.append(
        (jumlah, average_image, psnr)
    )

# Menampilkan gambar hasil
for jumlah, image, psnr in hasil:

    print(f'\nJumlah citra di-average : {jumlah}')
    print(f'Nilai PSNR              : {psnr:.2f} dB')

    cv2_imshow(image)

# Tabel
print('\nTABEL HASIL')
print('-----------------------------------------------')
print('No   Jumlah Citra       PSNR (dB)')
print('-----------------------------------------------')

for i, (jumlah, image, psnr) in enumerate(hasil, start=1):
    print(f'{i:<4} {jumlah:<18} {psnr:.2f}')

print('-----------------------------------------------')
```

```python
import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow

# ==========================================
# 1. MEMBACA GAMBAR ASLI
# ==========================================

original = cv.imread(
    '/content/drive/MyDrive/Colab Notebooks/rumah.png'
)

if original is None:
    print("Gambar couple.tiff tidak ditemukan.")
else:
    print("Gambar berhasil dibaca.")

    height, width = original.shape[:2]

    # ======================================
    # 2. MEMBUAT MASK
    # ======================================

    # Mask hitam
    mask = np.zeros((height, width), dtype=np.uint8)

    # Dua lingkaran putih
    # Sesuaikan posisi dengan gambar kamu
    cv.circle(mask, (int(width * 0.35), int(height * 0.28)),
              int(height * 0.13), 255, -1)

    cv.circle(mask, (int(width * 0.65), int(height * 0.28)),
              int(height * 0.13), 255, -1)

    # Ubah mask grayscale menjadi 3 channel
    mask_3ch = cv.cvtColor(mask, cv.COLOR_GRAY2BGR)


    # ======================================
    # 3. OPERASI BITWISE
    # ======================================

    # NOT
    hasil_not = cv.bitwise_not(mask_3ch)

    # OR
    hasil_or = cv.bitwise_or(original, mask_3ch)

    # AND
    hasil_and = cv.bitwise_and(original, mask_3ch)

    # NAND
    hasil_nand = cv.bitwise_not(
        cv.bitwise_and(original, mask_3ch)
    )

    # XOR
    hasil_xor = cv.bitwise_xor(original, mask_3ch)


    # ======================================
    # 4. MENAMPILKAN HASIL
    # ======================================

    print("\nMASK")
    cv2_imshow(mask)

    print("\n1. NOT (Komplemen)")
    cv2_imshow(hasil_not)

    print("\n2. OR (Atau)")
    cv2_imshow(hasil_or)

    print("\n3. AND (Dan)")
    cv2_imshow(hasil_and)

    print("\n4. NAND (Not And)")
    cv2_imshow(hasil_nand)

    print("\n5. XOR (Exclusive Or)")
    cv2_imshow(hasil_xor)
```

```python
import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow

# ==========================================
# Membaca foto malam
# ==========================================

original = cv.imread(
    '/content/drive/MyDrive/Colab Notebooks/rumah.png'
)

if original is None:
    print('Foto tidak ditemukan. Periksa path gambar.')
else:

    # ======================================
    # Gamma Correction
    # ======================================

    gamma = 0.5

    gamma_image = np.zeros_like(
        original,
        dtype=np.uint8
    )

    for y in range(original.shape[0]):
        for x in range(original.shape[1]):
            for c in range(original.shape[2]):

                pixel = original[y, x, c]

                # Normalisasi pixel 0-1
                normalized = pixel / 255.0

                # Rumus Gamma Correction
                corrected = 255 * (normalized ** gamma)

                gamma_image[y, x, c] = np.clip(
                    corrected,
                    0,
                    255
                )

    # ======================================
    # Menampilkan hasil
    # ======================================

    print('Foto asli:')
    cv2_imshow(original)

    print('Hasil Gamma Correction (γ = 0.5):')
    cv2_imshow(gamma_image)
```

```python
import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow

# ==========================================
# 1. Membaca gambar
# ==========================================

original = cv.imread(
    '/content/drive/MyDrive/Colab Notebooks/rumah.png'
)

if original is None:
    print('Gambar crayfish.jpg tidak ditemukan.')
else:
    print('Gambar berhasil dibaca.')

    # ======================================
    # 2. GAMMA CORRECTION
    # ======================================

    gamma = 0.6

    # Membuat lookup table gamma
    table = np.array([
        255 * ((i / 255.0) ** gamma)
        for i in np.arange(0, 256)
    ]).astype(np.uint8)

    gamma_image = cv.LUT(original, table)


    # ======================================
    # 3. CONTRAST DAN BRIGHTNESS
    # ======================================

    alpha = 1.8   # contrast
    beta = 10     # brightness

    enhanced = cv.convertScaleAbs(
        gamma_image,
        alpha=alpha,
        beta=beta
    )


    # ======================================
    # 4. MENAMPILKAN BEFORE - AFTER
    # ======================================

    print('Citra Asli (Before)')
    cv2_imshow(original)

    print('Citra Hasil Enhancement (After)')
    cv2_imshow(enhanced)


    # ======================================
    # 5. Menampilkan berdampingan
    # ======================================

    before_after = cv.hconcat(
        (original, enhanced)
    )

    print('Before | After')
    cv2_imshow(before_after)
```
