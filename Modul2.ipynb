```python
import cv2 as cv
from google.colab.patches import cv2_imshow
import numpy as np

# Membaca gambar
img = cv.imread('/content/drive/MyDrive/Google Collab/Citra/Modul2/kucing.webp')

# Mendapatkan ukuran gambar
tinggi, lebar = img.shape[:2]

# Menentukan tinggi bar secara adaptif
tinggi_bar = tinggi // 12

# Membuat salinan gambar
output = img.copy()

# Membuat bar hitam di bagian bawah gambar
cv.rectangle(
    output,
    pt1=(0, tinggi - tinggi_bar),
    pt2=(lebar, tinggi),
    color=(0, 0, 0),
    thickness=-1
)

# Menentukan ukuran font secara adaptif
font = cv.FONT_HERSHEY_SIMPLEX
font_scale = tinggi / 600
ketebalan = max(1, int(tinggi / 300))

# Teks identitas
teks = "Dokumentasi PCVK - 2026"

# Menentukan posisi teks secara adaptif
x_teks = int(lebar * 0.02)
y_teks = int(tinggi - tinggi_bar * 0.35)

# Menambahkan teks
cv.putText(
    output,
    teks,
    (x_teks, y_teks),
    font,
    font_scale,
    (255, 255, 255),
    ketebalan,
    cv.LINE_AA
)

# Menampilkan hasil
cv2_imshow(output)
```

```python
import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow

# Membaca gambar
img = cv.imread('/content/drive/MyDrive/Google Collab/Citra/Modul2/kucing.webp')

# Mendapatkan ukuran gambar
tinggi, lebar = img.shape[:2]

# Menentukan ukuran kanvas persegi
sisi = max(tinggi, lebar)

# Membuat kanvas berwarna putih
canvas = np.ones((sisi, sisi, 3), dtype=np.uint8) * 255

# Menentukan ukuran gambar setelah resize dengan rasio tetap
if lebar > tinggi:
    lebar_baru = sisi
    tinggi_baru = int(tinggi * sisi / lebar)
else:
    tinggi_baru = sisi
    lebar_baru = int(lebar * sisi / tinggi)

# Resize gambar tanpa mengubah rasio
img_resize = cv.resize(img, (lebar_baru, tinggi_baru))

# Menentukan posisi agar gambar tepat di tengah
x = (sisi - lebar_baru) // 2
y = (sisi - tinggi_baru) // 2

# Menempatkan gambar pada kanvas
canvas[y:y+tinggi_baru, x:x+lebar_baru] = img_resize

# Menentukan ukuran font secara proporsional
font = cv.FONT_HERSHEY_SIMPLEX
font_scale = sisi / 500
ketebalan = max(1, int(sisi / 250))

# Nama toko
teks = "TOKO KUCING"

# Menghitung ukuran teks
(text_width, text_height), _ = cv.getTextSize(
    teks, font, font_scale, ketebalan
)

# Posisi teks di bagian bawah tengah
text_x = (sisi - text_width) // 2
text_y = sisi - int(sisi * 0.05)

# Menambahkan teks
cv.putText(
    canvas,
    teks,
    (text_x, text_y),
    font,
    font_scale,
    (0, 0, 0),
    ketebalan,
    cv.LINE_AA
)

# Menampilkan hasil
cv2_imshow(canvas)
```
